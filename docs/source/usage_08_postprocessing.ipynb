# Usage | 8. Postprocessing
This notebook explores potential ways of postprocessing results from GrowBikeNet.

## Network analysis

[The GeoDataFrame](https://docs.bikenetkit.org/GrowBikeNet/usage_02_network_growth/) produced by GrowBikeNet, or the [exported data](https://docs.bikenetkit.org/GrowBikeNet/usage_04_data_export_and_visualization/), has the form of edges with a spatial geometry. It can be further analyzed with or without its spatial aspects, see below.

### Spatial network analysis

GrowBikeNet's resulting data can be read by spatial network analysis libraries such as [pandarm](https://github.com/oturns/pandarm) for fast shortest path or isochrone/service area analysis, or [spaghetti](https://pysal.org/spaghetti/notebooks/quickstart.html) for network-constrained analysis.

GrowBikeNet's resulting data can be read by GIS software like [QGIS](https://qgis.org/) and analyzed by its [built-in network analysis functionalities](https://docs.qgis.org/3.44/en/docs/training_manual/vector_analysis/network_analysis.html) or plugins like [QNEAT](https://root676.github.io/), also for shortest path or isochrone/service area analysis.

### Non-spatial network analysis

GrowBikeNet's resulting data can be transformed into a NetworkX Graph, as explained in the NetworkX documentation: [https://networkx.org/documentation/stable/auto_examples/geospatial/plot_lines.html](https://networkx.org/documentation/stable/auto_examples/geospatial/plot_lines.html)  
It is beyond the scope of our documentation to go into the detailed capabilities of NetworkX.
