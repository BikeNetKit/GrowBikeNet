# Minimum working example
This notebook runs a minimum working example with growbikenet.

## Import growbikenet

The standard way of importing growbikenet:

```python
import growbikenet as gbn
```

## Run growbikenet on example city

```python
edges_ordered = gbn.growbikenet("Lyon")
```

## Explore results

See the contents of the resulting GeoDataFrame:

```python
edges_ordered.head()
```

The columns of this GeoDataFrame are explained in the [Usage docs](https://docs.bikenetkit.org/GrowBikeNet/usage_02_network_growth/).

Explore the results interactively:

```python
edges_ordered.explore( 
    tiles="https://tiles.stadiamaps.com/tiles/alidade_smooth/{z}/{x}/{y}{r}.png",
    attr="(C) Stadia Maps (C) OpenMapTiles (C) OpenStreetMap contributors",
    style_kwds={"weight": 3, "color": "#096a51"},
)
```

The `explore()` method is built-in to GeoDataFrames. To export plots, see the [Usage docs](https://docs.bikenetkit.org/GrowBikeNet/usage_04_data_export_and_visualization/#generating-plots).
